```python
name = []
domain_location = []
rating = []
review_count = []
salaries = []
interviews = []
jobs = []
benefits = []
Highlyratedfor=[]
Criticallyratedfor=[]

for i in company:
    name.append(i.find('h2').text.strip())
    rating.append(i.find('span',class_='text-primary-text font-pn-700 text-sm !text-[14px] !leading-[14px]').text)
    domain_location.append(i.find('span',class_='overflow-hidden text-left text-[14px] font-pn-400 leading-5 tracking-[0.25px] text-[#394468] max-md:text-[12px] max-md:leading-4').text)
    hrf = None
    crf = None

    labels = i.find_all('span')

    for label in labels:
        text = label.get_text(strip=True).upper()

        if "HIGHLY RATED FOR" in text:
            value = label.find_next('span', class_='text-left text-[14px] font-pn-400 leading-5 tracking-[0.25px] text-[#1E223C] max-md:text-[12px] max-md:leading-4')
            hrf = value.get_text(strip=True) if value else None

        elif "CRITICALLY RATED FOR" in text:
            value = label.find_next('span', class_='text-left text-[14px] font-pn-400 leading-5 tracking-[0.25px] text-[#1E223C] max-md:text-[12px] max-md:leading-4')
            crf = value.get_text(strip=True) if value else None

    Highlyratedfor.append(hrf)
    Criticallyratedfor.append(crf)
    
for j in company_info:
    review_count.append(j.find_all('span',class_='text-center text-[16px] font-pn-700 leading-6 tracking-[0.15px] text-[#2A2F4F] max-md:text-[14px] max-md:font-pn-600')[0].text)
    salaries.append(j.find_all('span',class_='text-center text-[16px] font-pn-700 leading-6 tracking-[0.15px] text-[#2A2F4F] max-md:text-[14px] max-md:font-pn-600')[1].text)
    interviews.append(j.find_all('span',class_='text-center text-[16px] font-pn-700 leading-6 tracking-[0.15px] text-[#2A2F4F] max-md:text-[14px] max-md:font-pn-600')[2].text)
    jobs.append(j.find_all('span',class_='text-center text-[16px] font-pn-700 leading-6 tracking-[0.15px] text-[#2A2F4F] max-md:text-[14px] max-md:font-pn-600')[3].text)
    benefits.append(j.find_all('span',class_='text-center text-[16px] font-pn-700 leading-6 tracking-[0.15px] text-[#2A2F4F] max-md:text-[14px] max-md:font-pn-600')[4].text)

data = {'name':name,'domain_and_location':domain_location,'rating':rating,'reviews':review_count,'salaries':salaries,'interviews':interviews,'jobs':jobs,'benefits':benefits,'Highlyratedfor':Highlyratedfor,'Criticallyratedfor':Criticallyratedfor}

df = pd.DataFrame(data)
```

```python
final = pd.DataFrame()
for k in range(1, 300):
    url='https://www.ambitionbox.com/list-of-companies?page={}'.format(k)
    headers={'User-Agent':'Mozilla/5.0 (Windows NT 6.3; Win 64 ; x64) Apple WeKit /537.36(KHTML , like Gecko) Chrome/80.0.3987.162 Safari/537.36'}
    webpage=requests.get(url,headers=headers).text
    soup = BeautifulSoup(webpage, 'lxml')
    company = soup.find_all('div', class_='flex cursor-pointer px-5 pt-5 gap-2 max-md:px-3 max-md:pt-4')
    company_info = soup.find_all('div', class_='flex justify-evenly pl-[55px] pr-5 max-md:justify-start max-md:gap-[25px] max-md:overflow-x-scroll max-md:px-3 max-md:pl-[60px] max-md:[scrollbar-width:none] max-md:[&::-webkit-scrollbar]:hidden')
    name = []
    domain_location = []
    rating = []
    review_count = []
    salaries = []
    interviews = []
    jobs = []
    benefits = []
    Highlyratedfor=[]
    Criticallyratedfor=[]

    for i in company:
        try:
            name.append(i.find('h2').text.strip())
        except:
            name.append(np.nan)
        
        try:
            rating.append(i.find('span',class_='text-primary-text font-pn-700 text-sm !text-[14px] !leading-[14px]').text)
        except:
            rating.append(np.nan)
        
        try:
            domain_location.append(i.find('span',class_='overflow-hidden text-left text-[14px] font-pn-400 leading-5 tracking-[0.25px] text-[#394468] max-md:text-[12px] max-md:leading-4').text)
        except:
            domain_location.append(np.nan)
        
        hrf = None
        crf = None

        labels = i.find_all('span')

        for label in labels:
            text = label.get_text(strip=True).upper()

            if "HIGHLY RATED FOR" in text:
                value = label.find_next('span', class_='text-left text-[14px] font-pn-400 leading-5 tracking-[0.25px] text-[#1E223C] max-md:text-[12px] max-md:leading-4')
                hrf = value.get_text(strip=True) if value else None

            elif "CRITICALLY RATED FOR" in text:
                value = label.find_next('span', class_='text-left text-[14px] font-pn-400 leading-5 tracking-[0.25px] text-[#1E223C] max-md:text-[12px] max-md:leading-4')
                crf = value.get_text(strip=True) if value else None

        Highlyratedfor.append(hrf)
        Criticallyratedfor.append(crf)

    for j in company_info:
        try:
            review_count.append(j.find_all('span',class_='text-center text-[16px] font-pn-700 leading-6 tracking-[0.15px] text-[#2A2F4F] max-md:text-[14px] max-md:font-pn-600')[0].text)
        except:
            review_count.append(np.nan)
        
        try:
            salaries.append(j.find_all('span',class_='text-center text-[16px] font-pn-700 leading-6 tracking-[0.15px] text-[#2A2F4F] max-md:text-[14px] max-md:font-pn-600')[1].text)
        except:
            salaries.append(np.nan)
        
        try:
            interviews.append(j.find_all('span',class_='text-center text-[16px] font-pn-700 leading-6 tracking-[0.15px] text-[#2A2F4F] max-md:text-[14px] max-md:font-pn-600')[2].text)
        except:
            interviews.append(np.nan)
        
        try:
            jobs.append(j.find_all('span',class_='text-center text-[16px] font-pn-700 leading-6 tracking-[0.15px] text-[#2A2F4F] max-md:text-[14px] max-md:font-pn-600')[3].text)
        except:
            jobs.append(np.nan)
        
        try:
            benefits.append(j.find_all('span',class_='text-center text-[16px] font-pn-700 leading-6 tracking-[0.15px] text-[#2A2F4F] max-md:text-[14px] max-md:font-pn-600')[4].text)
        except:
            benefits.append(np.nan)

    data = {'name':name,'domain_and_location':domain_location,'rating':rating,'reviews':review_count,'salaries':salaries,'interviews':interviews,'jobs':jobs,'benefits':benefits,'Highlyratedfor':Highlyratedfor,'Criticallyratedfor':Criticallyratedfor}
    df = pd.DataFrame(data)
    final = pd.concat([final, df], ignore_index=True)
```

```python
# spliting doamin and office count
# Extract Domain, Location, and Office Count using regex
final[['domain','location','office_count']] = final[
    'domain_and_location'].str.extract(r'^(.*?)\s*\|\s*(.*?)\s*\+(\d+)\s*other')

# Remove old combined column
final = final.drop(columns=['domain_and_location'])

# convert office_count into integer(int)
final['office_count'] = pd.to_numeric(final['office_count'])
```
